```python
import pandas as pd

# Load the dataset
file_path = 'boysnames.csv'  # Update with the correct file path
df = pd.read_csv(file_path)

# Clean the dataset by replacing '[x]' with NaN and converting 'count' to numeric
df['count'] = pd.to_numeric(df['count'], errors='coerce')

# Filter data for the name "Arthur"
arthur_data = df[df['name'] == 'Arthur']

# Group by year and calculate the total count for each year
arthur_summary = arthur_data.groupby('year')['count'].sum().reset_index()

# Sort by year for better visualization
arthur_summary.sort_values(by='year', inplace=True)
arthur_summary.reset_index(drop=True, inplace=True)

# Display the summarized data
print(arthur_summary)

# Optional: Plot the popularity trend over time
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(arthur_summary['year'], arthur_summary['count'], marker='o', label='Arthur')
plt.title('Popularity of the Name "Arthur" Over Time')
plt.xlabel('Year')
plt.ylabel('Count')
plt.grid(True)
plt.legend()
plt.show()
```
